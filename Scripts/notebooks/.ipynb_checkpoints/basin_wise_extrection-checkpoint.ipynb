```python
def covariance(x, y, dims=None):
    return xr.dot(x - x.mean(dims), y - y.mean(dims), dims=dims) / x.count(dims)

def corrrelation(x, y, dims=None):
    return covariance(x, y, dims) / (x.std(dims) * y.std(dims))

def bias(x, y, dims=None):
    return ((y - x).mean(dims))

def rmse(x, y, dims=None):
    return np.sqrt(((y - x) ** 2).mean(dims))

def ubrmse(x, y, dims=None):
    return np.sqrt(((rmse(x, y, dims))**2)-((bias(x, y, dims))**2))

def rrmse(x, y, dims=None):
    return rmse(x, y, dims=None)/y.mean(dims)

def pbias(x, y, dims=None):
    return (((y - x).sum(dims))/y.sum(dims))*100



def KGEbeta(x, y, dims=None):
    return  (x.mean(dims)/y.mean(dims))

def KGEalpha(x, y, dims=None):
    return  (x.std(dims)/y.std(dims))

def KGE(x, y, dims=None):
    return  (1- np.sqrt((1-corrrelation(x, y, dims))**2 + (1-(x.mean(dims)/y.mean(dims)))**2 + (1-(x.std(dims)/y.std(dims)))**2))
```
